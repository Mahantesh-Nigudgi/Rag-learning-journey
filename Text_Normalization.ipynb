```python
# loading and reading the data.tx file 

with open('data.txt','r',encoding='utf-8') as file:
    data=file.read()
```

# Text Normalization :-

```python
# Converting the whole data to lower case 

data=data.lower()
```

```python
# removing the extra space

import re
data=re.sub(r'\s{2,}'," ",data)
data
```

```python
# contracting the data 

import contractions
data = contractions.fix(data)
```

```python
# removibg all the punctuation from the data

import string
print(string.punctuation)
punc=string.punctuation

data=re.sub(f'[{punc}]',"",data)
```

```python
# removing all teh emoji from the data

import emoji

# data = emoji.demojize(data)
data=emoji.replace_emoji(data)
data
```

```python
# removing the special characters from data

data=re.sub(r'[^0-9a-zA-Z\s]','',data)
data
```

```python
# correcting all the spelling mistakes in the data using TextBlob

from textblob import TextBlob

text_data=TextBlob(data).correct()
data=str(text_data)
```

```python
import spacy
nlp=spacy.load('en_core_web_sm')
```

```python
# creating tokens

tokens=nlp(data)
updateed_tokens=[token.lemma_ for token in tokens if not token.is_stop]
data=' '.join(updateed_tokens)
```

```python
# creating the chunks

from langchain_text_splitters import RecursiveCharacterTextSplitter

text_spiltter=RecursiveCharacterTextSplitter(
    chunk_size=100,
    chunk_overlap=20,
    separators=['\n\n','\n']

)

chunks=text_spiltter.split_text(data)
```
